24. Print the list of all countries in the data.

```python
country_mapping = {
    'Usa': 'United States', 'Us': 'United States', 'U.S.A.': 'United States',
    'Uk': 'United Kingdom', 'U.K.': 'United Kingdom', 'Great Britain': 'United Kingdom'
}
df['client_country'] = df['client_country'].astype(str).str.strip().str.title().replace(country_mapping)
df['client_country'].unique()
```

25. Convert the hours logged column to a number.

```python
df['hours_logged'] = pd.to_numeric(df['hours_logged'].astype(str).str.replace('hrs', '', case=False).str.strip(), errors='coerce')
df['hours_logged'].head()
```

26. Convert the billing rate column to a number. Remove the $ sign first.

```python
df['billing_rate_(usd)'] = pd.to_numeric(df['billing_rate_(usd)'].astype(str).str.replace('$', '', regex=False).str.replace('USD', '', regex=False).str.strip(), errors='coerce')
df['billing_rate_(usd)'].head()
```

27. Convert the CSAT score column to a number.

```python
df['csat_score'] = pd.to_numeric(df['csat_score'], errors='coerce')
df['csat_score'].head()
```

28. Print the average, the smallest and the largest value of hours logged.

```python
pd.Series.describe(df['hours_logged'])
```

29. Sort the table by hours logged from highest to lowest and show the top 10 rows.

```python
df.sort_values(by='hours_logged', ascending=False).head(10)
```

30. Show the 5 tickets with the highest billable amount.

```python
df['billable_amount_usd'] = pd.to_numeric(df['billable_amount_usd'].astype(str).str.replace('$', '', regex=False).str.replace('USD', '', regex=False).str.replace(',', '', regex=False).str.strip(), errors='coerce')
df.sort_values(by='billable_amount_usd', ascending=False).head(5)
```

31. Remove the extra spaces from the client name column.

```python
df['client_name'] = df['client_name'].astype(str).str.strip()
df['client_name'].head(10)
```

32. Change the client name column to title case.

```python
df['client_name'] = df['client_name'].str.title()
df['client_name'].head(10)
```

33. Convert the ticket created column to a date.

```python
df['ticket_created_at'] = pd.to_datetime(df['ticket_created_at'], format='mixed', errors='coerce')
df['ticket_created_at'].head()
```

34. Add a new column that holds the year of the ticket.

```python
df['year'] = df['ticket_created_at'].dt.year
df[['ticket_id', 'ticket_created_at', 'year']].head()
```

35. Add a new column that holds the month of the ticket.

```python
df['month'] = df['ticket_created_at'].dt.month
df[['ticket_id', 'ticket_created_at', 'year', 'month']].head()
```

36. Count how many tickets were created in each year.

```python
df['year'].value_counts().sort_index()
```

37. Add a new column called total equal to billing rate multiplied by hours logged.

```python
df['total'] = df['billing_rate_(usd)'] * df['hours_logged']
df[['ticket_id', 'billing_rate_(usd)', 'hours_logged', 'total']].head()
```

38. Fill the missing values in the CSAT score column with the average CSAT score.

```python
mean_csat = df['csat_score'].mean()
df['csat_score'] = df['csat_score'].fillna(mean_csat)
print(f'Average CSAT: {mean_csat:.2f}')
print('Remaining missing CSAT values:', df['csat_score'].isna().sum())
```

39. Drop the rows where the client name is missing.

```python
before_count = len(df)
df = df[df['client_name'].notna() & (df['client_name'] != 'Nan') & (df['client_name'] != '')]
print(before_count - len(df))
len(df)
```

45. Draw a bar chart of ticket count by service line.

```python
import matplotlib.pyplot as plt

df['serviceline'].value_counts().plot(kind='bar', color='royalblue', edgecolor='black')

plt.title('Ticket Count by Service Line', fontsize=14, fontweight='bold')
plt.xlabel('Service Line', fontsize=12)
plt.ylabel('Number of Tickets', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

46. Draw a histogram of hours logged.

```python
plt.figure(figsize=(10, 5))
df[df['hours_logged'].between(0, 100)]['hours_logged'].plot(kind='hist', bins=30, color='teal', edgecolor='black')
plt.title('Distribution of Hours Logged (0-100 Hours)', fontsize=14, fontweight='bold')
plt.xlabel('Hours Logged', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
